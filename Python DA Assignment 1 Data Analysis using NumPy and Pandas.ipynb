```python
import numpy as np
```

```python
# 1. Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
# 2. Inspection and Properties
print("\nShape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# 3. Array Operations

# Celsius to Fahrenheit
fahrenheit = (temperatures_w1 * 9/5) + 32
print("\nTemperatures in Fahrenheit:")
print(fahrenheit)
```

```python
# Maximum, Minimum and Mean
print("\nMaximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python

# 4. Array Slicing and Indexing

# First three days
print("\nFirst Three Days:")
print(temperatures_w1[:3])
```

```python
# Weekend (last two days)
print("\nWeekend Temperatures:")
print(temperatures_w1[-2:])
```

```python
# Middle three days
print("\nMiddle Three Days:")
print(temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
    ])
```

```python
print("\n2D Temperature Array:")
print(temperatures)
```

```python
# 6. Inspect and Slice the 2D Array

print("\nShape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)
```

```python
# Temperatures for each week
print("\nWeek 1:")
print(temperatures[0])
```

```python
print("\nWeek 2:")
print(temperatures[1])
```

```python
# Weekend temperatures for both weeks
print("\nWeekend Temperatures for Both Weeks:")
print(temperatures[:, -2:])
```

```python
import pandas as pd
```

```python
# Create DataFrame

data = {
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109],
    "CustomerName": ["Asha", "Rahul", "Priya", "Kiran", "Deepa",
                     "Arun", "Sneha", "Vijay", "Meera"],
    "Region": ["North", "South", "East", "West", "North",
               "South", "East", "West", "North"],
    "ProductCategory": ["Electronics", "Clothing", "Electronics",
                        "Furniture", "Clothing", "Electronics",
                        "Furniture", "Clothing", "Electronics"],
    "Amount": [150, 300, 450, 200, 250, 300, 180, 350, 400]
}
```

```python
transactions = pd.DataFrame(data)
```

```python
# Data Exploration

print("DataFrame:")
print(transactions)

print("\nHead:")
print(transactions.head())

print("\nTail:")
print(transactions.tail())

print("\nShape:")
print(transactions.shape)

print("\nColumn Names:")
print(transactions.columns)

print("\nData Types:")
print(transactions.dtypes)

# Display ProductCategory and Amount
print("\nProductCategory and Amount:")
print(transactions[["ProductCategory", "Amount"]])

# Last 3 columns
print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])

# Region = North and Amount > 200
print("\nNorth Region and Amount > 200:")
print(transactions[
    (transactions["Region"] == "North") &
    (transactions["Amount"] > 200)
])
```

```python
# Value Counts
print("\nProduct Category Counts:")
print(transactions["ProductCategory"].value_counts())
```

```python
# Unique Regions
print("\nUnique Regions:")
print(transactions["Region"].unique())
```

```python
# Group by Region and Mean Amount
print("\nMean Amount by Region:")
print(transactions.groupby("Region")["Amount"].mean())
```

```python
# Data Manipulation
# Modify Amount for TransactionID 102
transactions.loc[
    transactions["TransactionID"] == 102,
    "Amount"
] = 165
```

```python
# Add Discount Column (10%)
transactions["Discount"] = transactions["Amount"] * 0.10
```

```python
print("\nAfter Adding Discount:")
print(transactions)
```

```python
# Remove TransactionID 109
transactions = transactions[
    transactions["TransactionID"] != 109
]

print("\nAfter Removing TransactionID 109:")
print(transactions)
```

```python
# Delete Discount Column
transactions.drop("Discount", axis=1, inplace=True)

print("\nAfter Deleting Discount Column:")
print(transactions)
```

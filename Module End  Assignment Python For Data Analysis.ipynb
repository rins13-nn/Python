```python
import pandas as pd

df = pd.read_csv("social_media_engagement_5000.csv")

print("Dataset loaded successfully!")
```

```python
# Numeric columns → Median
df['age'] = df['age'].fillna(df['age'].median())
df['likes'] = df['likes'].fillna(df['likes'].median())
df['comments'] = df['comments'].fillna(df['comments'].median())
df['shares'] = df['shares'].fillna(df['shares'].median())

# Categorical columns → Mode
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
df['sentiment'] = df['sentiment'].fillna(df['sentiment'].mode()[0])

print("Missing values handled successfully!")
```

```python
df.isnull().sum()
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates removed successfully!")
print("New Shape:", df.shape)
```

```python
df['posted_at'] = pd.to_datetime(df['posted_at'])

print("Date conversion successful!")
```

```python
df['posted_at'].head(10)
```

```python
print(df['posted_at'].unique()[:10])
```

```python
df['posted_at'] = pd.to_datetime(
    df['posted_at'],
    format='mixed',
    dayfirst=True
)

print("Date conversion successful!")
```

```python
print(df['posted_at'].dtype)
```

```python
df['hashtag_count'] = df['hashtags'].apply(
    lambda x: len(str(x).split())
)

df[['hashtags', 'hashtag_count']].head()
```

```python
df['engagement_score'] = (
    df['likes'] +
    df['comments'] +
    df['shares']
)

df[['likes', 'comments', 'shares', 'engagement_score']].head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
corr_matrix = df.select_dtypes(include=np.number).corr()

corr_matrix
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['impression_count'],
    df['likes'],
    alpha=0.6
)

plt.xlabel("Impressions")
plt.ylabel("Likes")
plt.title("Likes vs Impressions")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    df['age'],
    bins=20,
    color='skyblue',
    edgecolor='black'
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
gender_counts = df['gender'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%'
)

plt.title("Gender Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.boxplot(df['engagement_rate'])

plt.title("Engagement Rate Distribution")

plt.show()
```

```python
daily_engagement = df.groupby(
    df['posted_at'].dt.date
)['engagement_score'].mean()

plt.figure(figsize=(12,5))

plt.plot(
    daily_engagement.index,
    daily_engagement.values
)

plt.title("Daily Engagement Trend")
plt.xlabel("Date")
plt.ylabel("Average Engagement Score")

plt.xticks(rotation=45)

plt.show()
```

```python
df['post_category'].value_counts().plot(
    kind='bar',
    figsize=(8,5),
    color='orange'
)

plt.title("Posts by Category")
plt.xlabel("Category")
plt.ylabel("Count")

plt.show()
```

```python
sample_df = df.sample(300, random_state=42)

sns.pairplot(
    sample_df[
        ['likes',
         'comments',
         'shares',
         'engagement_score',
         'sentiment']
    ],
    hue='sentiment'
)

plt.show()
```

```python
sample_df = df.sample(500, random_state=42)

plt.figure(figsize=(10,5))

sns.swarmplot(
    data=sample_df,
    x='device_type',
    y='engagement_score'
)

plt.title("Engagement vs Device Type")

plt.show()
```
